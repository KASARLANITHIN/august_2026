```python
x=np.array([40,40,30,20,10])
y=np.array([12,34,22,19,40])
plt.plot(x,y)
plt.xlabel("high")
plt.ylabel("low")
plt.title("Visualization")
plt.grid()
plt.show()
```

```python
x=np.array([40,40,30,20,10])
y=np.array([12,34,22,19,40])
plt.plot(x,y,"D")
plt.xlabel("high")
plt.ylabel("low")
plt.title("Visualization")
plt.grid()
plt.show()
```

```python
x=np.array([40,40,30,20,10])
y=np.array([6,45,12,30,22])
plt.plot(x,y,color='g')
plt.xlabel("high")
plt.ylabel("low")
plt.title("Visualization")
plt.grid()
plt.show()
```

```python
x=np.array([40,40,30,20,10])
y=np.array([6,45,12,30,22])
plt.plot(x,y)
plt.plot(x,y,color='r',marker="P",ms=8,mec="b",linewidth=5)
plt.xlabel("high")
plt.ylabel("low")
plt.show()
```

```python
x=np.array([40,40,30,20,10])
y=np.array([6,45,12,30,22])
plt.plot(x,y)
plt.plot(x,y,color='b',marker="P",ms=4,mec="b",linewidth=3,linestyle="dashdot")
plt.xlabel("high")
plt.ylabel("low")
plt.show()
```
