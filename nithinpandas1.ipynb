```python
import pandas as pd
data={'name':['nithin','mahesh','sanvi','basheer','ramya'],
      'gender':['m','m','f','m','f'],
      'city':['hyd','hyd','nrml','nzb','nzb']}
df=pd.DataFrame(data,columns=['name','gender','city'])
print(df)
```
